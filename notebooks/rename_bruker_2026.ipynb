```python
from usnm2p.bruker_utils import *
from usnm2p.constants import *
from usnm2p.logger import logger
from usnm2p.fileops import get_data_root
```

### Inputs

```python
# Dataset parameters
analysis = 'main'
mouseline = 'theo_line3'
expdate = '20260928'
mouseid = 'DT1684'
region = 'region1'
layer = DEFAULT_LAYER

# Acquisition parameters
fs = 3.58  # acquisition frame rate (Hz)
npertrial = 100  # number of frames per trial

logmsg = '\n'.join([
    f'  - analysis: {analysis}',
    f'  - mouse line: {mouseline}',
    f'  - experiment date: {expdate}',
    f'  - mouse ID: {mouseid}',
    f'  - region: {region}',
    f'  - layer: {layer}',
    f'  - acquisition frame rate: {fs} Hz',
    f'  - number of frames per trial: {npertrial}',
])
logger.info(f'current dataset:\n{logmsg}')
```

### Get data directory for dataset of interest

```python
dataroot = get_data_root(kind=DataRoot.RAW_BRUKER)
logger.info(f'data root: {dataroot}')

# Construct dataset ID
dataset_id = f'{expdate}_{mouseid}_{region}'
if layer != DEFAULT_LAYER:
    dataset_id = f'{dataset_id}_{layer}'

# Extract input data directory
datadir = os.path.join(dataroot, analysis, mouseline, dataset_id)
if not os.path.exists(datadir):
    raise ValueError(f'input data directory "{datadir}" does not exist')
logger.info(f'input Bruker folder: {datadir}')
```

### Perform recursive files and subfolders renaming in experiment folder

```python
rename_bruker_2026_tifs(datadir, mouseline, npertrial, fs)
```
